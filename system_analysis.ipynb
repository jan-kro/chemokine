```python
import utils
import h5py
import os
import ffmpeg
import numpy as np
```

```python
# system to analyse
dir_sys = "/net/data05/janmak98/results/september/chemokine/tests/"
name_sys = "test_tracking"
```

```python
# load trajectory
fname_particles = utils.get_fname(dir_sys, name_sys, "particle_traj")
print("Loading trajectory from", fname_particles)

fh5 = h5py.File(fname_particles, "r")
print("\nKeys in trajectory file:", list(fh5.keys()))

traj = fh5["trajectory"]

print("\nfh5[\'trajectory\'] metadata:")
print("Trajectory shape:                   ", traj.shape)
print("Trajectory attributes:              ", traj.attrs.keys())
print("Tracked particle type logical index:", traj.attrs["track_types"])

track_types = traj.attrs["track_types"]
ptypes = np.arange(len(track_types))
print("Tracked article types:              ", ptypes[track_types])
```

```python
# load config file
fname_config = utils.get_fname(dir_sys, name_sys, "config")
print("\nLoading config from", fname_config)
config = utils.load_config_dict(fname_config)

print("\nKeys in config file:\n", list(config.keys()))
print("\nConfig comment (system information):\n", config["comment"])
print("\nParticle names:", config["particle_names"])
print("\nNames of tracked particles:", np.array(config["particle_names"])[track_types])
```

```python
# animate CCL5 trajectory
traj_ccl5 = traj[:,:,:,0]

# by indexing the "trajectory" dataset, the data is loaded into memory
# now the file can be closed

# close fh5 file
fh5.close()

# replace every positive number with 1 
traj_ccl5_untracked = np.array(traj_ccl5!=0, dtype=int)

fname_anim_ccl5_untracked = utils.get_fname(dir_sys, name_sys+"_ccl5", "animation")

if not os.path.exists(fname_anim_ccl5_untracked):
    utils.animate_ffmpeg(traj_ccl5_untracked, fname=fname_anim_ccl5_untracked, show_collagen=False)
```

```python
# animate full trajectory

fname_traj = utils.get_fname(dir_sys, name_sys, "trajectory")
fname_anim = utils.get_fname(dir_sys, name_sys, "animation")

with h5py.File(fname_traj, "r") as fh5:
    if not os.path.exists(fname_anim):
        utils.animate_ffmpeg(fh5["trajectory"], fname=fname_anim, show_collagen=False)
        
```
